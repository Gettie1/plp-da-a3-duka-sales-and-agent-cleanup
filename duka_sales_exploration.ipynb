## 2. Data Preparation and Feature Engineering

Five derived columns are created to support revenue analysis,
monthly comparisons, weekday and weekend comparisons, and product
price segmentation.

```python

# 1. Calculate revenue
sales["revenue"] = sales["quantity"] * sales["unit_price"]

# 2. Extract month
sales["month"] = sales["order_date"].dt.month_name()

# 3. Extract weekday
sales["weekday"] = sales["order_date"].dt.day_name()

# 4. Identify weekend orders
sales["is_weekend"] = sales["order_date"].dt.day_of_week >= 5

# 5. Categorise products by price
sales["price_band"] = np.where(
    sales["unit_price"] >= 28000,
    "Premium",
    "Value"
)

# Display updated dataset
sales.head()
```

```python

# Calculate total revenue for each branch
branch_revenue = (
    sales.groupby("branch")["revenue"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

# Add revenue ranking
branch_revenue["revenue_rank"] = range(
    1, len(branch_revenue) + 1
)

print("BRANCH RANKING BY TOTAL REVENUE")
display(branch_revenue)
```

```python

# Count orders received by each branch
branch_orders = (
    sales.groupby("branch")["order_id"]
    .count()
    .sort_values(ascending=False)
    .reset_index(name="number_of_orders")
)

# Add order ranking
branch_orders["order_rank"] = range(
    1, len(branch_orders) + 1
)

print("BRANCH RANKING BY NUMBER OF ORDERS")
display(branch_orders)
```

```python

# Merge revenue and order rankings
branch_comparison = pd.merge(
    branch_revenue,
    branch_orders,
    on="branch"
)

# Check whether both rankings are identical
branch_comparison["same_rank"] = (
    branch_comparison["revenue_rank"]
    == branch_comparison["order_rank"]
)

display(branch_comparison)
```

```python

# Calculate average revenue per order for each branch
branch_comparison["average_revenue_per_order"] = (
    branch_comparison["branch"].map(
        sales.groupby("branch")["revenue"].mean()
    )
)

display(
    branch_comparison[
        [
            "branch",
            "revenue",
            "number_of_orders",
            "average_revenue_per_order",
            "revenue_rank",
            "order_rank",
            "same_rank"
        ]
    ]
)
```

The rankings were **the same** because revenue

A branch may receive many orders but generate less revenue if

Similarly, a branch with fewer orders may generate higher revenue

## Task 4: Top Three Products by Revenue and Units Sold

This section identifies the three best-performing products based on total revenue generated and total units sold.

The two rankings are compared to understand whether the products selling the highest number of units also generate the highest revenue.

```python

# Group products and calculate total revenue
product_revenue = (
    sales.groupby("product")
    .agg(
        total_revenue=("revenue", "sum"),
        units_sold=("quantity", "sum")
    )
    .sort_values(
        by="total_revenue",
        ascending=False
    )
)

# Select the top three products
top_three_revenue = product_revenue.head(3)

print("TOP THREE PRODUCTS BY REVENUE")
display(top_three_revenue)
```

```python

# Group products and calculate total units sold
product_units = (
    sales.groupby("product")
    .agg(
        units_sold=("quantity", "sum"),
        total_revenue=("revenue", "sum")
    )
    .sort_values(
        by="units_sold",
        ascending=False
    )
)

# Select the top three products
top_three_units = product_units.head(3)

print("TOP THREE PRODUCTS BY UNITS SOLD")
display(top_three_units)
```

```python

# Combine both rankings
product_comparison = pd.DataFrame({
    "revenue_rank": product_revenue.index,
    "total_revenue": product_revenue["total_revenue"].values,
    "units_rank": product_units.index,
    "total_units_sold": product_units["units_sold"].values
})

print("PRODUCT PERFORMANCE COMPARISON")
display(product_comparison)
```

### Task 4: Findings and Interpretation

The product analysis reveals that laptops generated the highest revenue, contributing **KES 41,535,000** from **639 units sold**. Tablets ranked second in revenue, generating **KES 19,348,000** from **691 units sold**, while phones ranked third, generating **KES 8,880,000** from **592 units sold**.

The ranking by revenue differs from the ranking by units sold. Tablets were the most purchased product, while laptops generated the most revenue.

This difference is mainly explained by the variation in unit prices. A laptop costs KES 65,000, compared to KES 28,000 for a tablet. Therefore, even though fewer laptops were sold, their higher selling price resulted in greater total revenue.

Management should maintain sufficient laptop inventory because of its strong revenue contribution while continuing to monitor tablet demand due to its higher sales volume.

## Task 5: Weekend vs Weekday Sales Analysis

This section compares sales performance between weekdays and weekends.

The analysis considers the number of orders, total revenue and average revenue per order to determine how purchasing patterns differ between the two periods.

```python

# Group sales by weekend and weekday
weekend_analysis = (
    sales.groupby("is_weekend")
    .agg(
        number_of_orders=("order_id", "count"),
        total_revenue=("revenue", "sum"),
        average_order_revenue=("revenue", "mean")
    )
    .reset_index()
)

# Replace Boolean values with readable labels
weekend_analysis["is_weekend"] = weekend_analysis[
    "is_weekend"
].map({
    False: "Weekday",
    True: "Weekend"
})

# Display results
display(weekend_analysis.round(2))
```

```python
# Extract average revenues
weekday_avg = weekend_analysis.loc[
    weekend_analysis["is_weekend"] == "Weekday",
    "average_order_revenue"
].iloc[0]

weekend_avg = weekend_analysis.loc[
    weekend_analysis["is_weekend"] == "Weekend",
    "average_order_revenue"
].iloc[0]

# Calculate percentage difference
percentage_difference = (
    (weekend_avg - weekday_avg) / weekday_avg
) * 100

print(
    f"Weekend average order revenue is "
    f"{percentage_difference:.2f}% higher than weekday average."
)
```

### Task 5: Findings and Interpretation

The analysis shows that Duka Electronics received **723 weekday orders** and **277 weekend orders**.

Weekday sales generated the highest total revenue of **KES 51,856,200**, compared to **KES 19,936,300** generated during weekends. This is mainly because weekday orders accounted for 72.3% of all orders.

However, weekend orders recorded a slightly higher average revenue per order of **KES 71,972.20**, compared to **KES 71,723.65** on weekdays. This represents a difference of approximately 0.35%.

These findings suggest that weekdays contribute more to overall revenue because of their higher order volume, while weekend customers spend slightly more per order on average.

Management should maintain strong weekday operations while exploring weekend promotions and marketing strategies to increase weekend order volume.

## Task 6: Best and Worst Revenue Months

This section analyses Duka Electronics' monthly revenue performance
from January to June 2026.

The objective is to identify the month with the highest revenue,
the month with the lowest revenue and calculate the percentage
difference between their revenue totals.

```python

# Group sales by month and calculate total revenue
monthly_revenue = (
    sales.groupby(sales["order_date"].dt.month)
    .agg(
        total_revenue=("revenue", "sum"),
        number_of_orders=("order_id", "count")
    )
)

# Convert month numbers into month names
monthly_revenue.index = pd.Index(
    monthly_revenue.index
).map(lambda month: pd.Timestamp(
    year=2026, month=int(month), day=1
).month_name())

# Display monthly revenue
print("MONTHLY REVENUE ANALYSIS")
display(monthly_revenue)
```

```python

# Identify the month with the highest revenue
best_month = monthly_revenue["total_revenue"].idxmax()
best_revenue = monthly_revenue["total_revenue"].max()

# Identify the month with the lowest revenue
worst_month = monthly_revenue["total_revenue"].idxmin()
worst_revenue = monthly_revenue["total_revenue"].min()

# Display findings
print("BEST PERFORMING MONTH")
print("Month:", best_month)
print("Revenue: KES", f"{best_revenue:,.0f}")

print("\nWORST PERFORMING MONTH")
print("Month:", worst_month)
print("Revenue: KES", f"{worst_revenue:,.0f}")
```

### Task 6: Findings and Interpretation

The monthly revenue analysis shows that April was the best-performing
month, generating KES 13,518,300 from 167 orders.

June was the worst-performing month, generating KES 10,903,400
from 162 orders.

The percentage difference between the best and worst months was
23.98%, meaning April generated approximately 23.98% more revenue
than June.

This difference suggests that Duka Electronics experienced
variations in sales performance throughout the first half of 2026.

Management should investigate the factors that contributed to
April's strong performance and identify possible reasons for
June's lower revenue. These findings can help improve future
sales planning, inventory management and promotional strategies.

## Task 7: Premium Online Orders from Nairobi and Mombasa

This section identifies Premium orders placed through the Online
sales channel in Nairobi and Mombasa branches.

The analysis calculates the number of qualifying orders and their
total revenue. The filtered records are exported to a CSV file

```python
premium_online_orders = sales[
    (sales["price_band"] == "Premium") &
    (sales["channel"] == "Online")&
    (sales["branch"].isin(["Nairobi", "Mombasa"]))
].copy()
display(premium_online_orders.head())
```

```python
# Count qualifying orders
total_premium_online_orders = len(premium_online_orders)

print(
    "Total Premium online orders from Nairobi and Mombasa:",
    total_premium_online_orders
)
```

```python
# Calculate total revenue
premium_online_revenue = premium_online_orders["revenue"].sum()

print(
    "Total revenue: KES",
    f"{premium_online_revenue:,.2f}"
)
```

```python
# Summarise Premium online sales by branch
premium_branch_summary = (
    premium_online_orders.groupby("branch")
    .agg(
        number_of_orders=("order_id", "count"),
        total_revenue=("revenue", "sum")
    )
    .sort_values(
        by="total_revenue",
        ascending=False
    )
)

display(premium_branch_summary)
```

```python

from pathlib import Path
import pandas as pd
from openpyxl import load_workbook
from openpyxl.styles import Font, PatternFill, Alignment
from openpyxl.worksheet.table import Table, TableStyleInfo

# Create the Excel output folder
Path("excel").mkdir(parents=True, exist_ok=True)

# Make sure revenue exists
if "revenue" not in sales.columns:
    sales["revenue"] = sales["quantity"] * sales["unit_price"]

# Branch and region mapping
region_lookup = pd.DataFrame({
    "branch": ["Nairobi", "Mombasa", "Kisumu", "Nakuru", "Eldoret"],
    "region": [
        "Capital Region",
        "Coast Region",
        "Western Region",
        "Rift Valley Region",
        "Rift Valley Region"
    ]
})

# Branch summary
branches = sorted(sales["branch"].unique())

branch_summary = pd.DataFrame({
    "branch": branches
})

# Create workbook with sales data and summary
file_path = "excel/duka_branch_summary.xlsx"

with pd.ExcelWriter(file_path, engine="openpyxl") as writer:
    sales.to_excel(writer, sheet_name="Sales Data", index=False)
    branch_summary.to_excel(writer, sheet_name="Branch Summary", index=False)
    region_lookup.to_excel(writer, sheet_name="Region Lookup", index=False)

# Open workbook to add formulas and formatting
wb = load_workbook(file_path)

ws_sales = wb["Sales Data"]
ws_summary = wb["Branch Summary"]
ws_lookup = wb["Region Lookup"]

# Create Excel tables
sales_table = Table(
    displayName="SalesData",
    ref=f"A1:{ws_sales.cell(1, ws_sales.max_column).column_letter}{ws_sales.max_row}"
)
sales_table.tableStyleInfo = TableStyleInfo(
    name="TableStyleMedium2",
    showRowStripes=True
)
ws_sales.add_table(sales_table)

lookup_table = Table(
    displayName="RegionLookup",
    ref=f"A1:B{ws_lookup.max_row}"
)
lookup_table.tableStyleInfo = TableStyleInfo(
    name="TableStyleMedium4",
    showRowStripes=True
)
ws_lookup.add_table(lookup_table)

# Add summary headings
ws_summary["B1"] = "Total Revenue (KES)"
ws_summary["C1"] = "Number of Orders"
ws_summary["D1"] = "Region"

# Add formulas for each branch
for row in range(2, ws_summary.max_row + 1):
    ws_summary[f"B{row}"] = (
        f'=SUMIFS(SalesData[revenue],SalesData[branch],A{row})'
    )
    ws_summary[f"C{row}"] = (
        f'=COUNTIFS(SalesData[branch],A{row})'
    )
    ws_summary[f"D{row}"] = (
        f'=XLOOKUP(A{row},RegionLookup[branch],RegionLookup[region],"Not found")'
    )

# Add grand total
total_row = ws_summary.max_row + 1
ws_summary[f"A{total_row}"] = "Grand Total"
ws_summary[f"B{total_row}"] = f"=SUM(B2:B{total_row - 1})"
ws_summary[f"C{total_row}"] = f"=SUM(C2:C{total_row - 1})"

# Format headings
for ws in [ws_sales, ws_summary, ws_lookup]:
    ws.freeze_panes = "A2"
    for cell in ws[1]:
        cell.font = Font(bold=True, color="FFFFFF")
        cell.fill = PatternFill(
            fill_type="solid",
            fgColor="1F4E78"
        )
        cell.alignment = Alignment(horizontal="center")

# Format revenue values
for row in range(2, total_row + 1):
    ws_summary[f"B{row}"].number_format = '"KES "#,##0.00'

# Set useful column widths
for ws in [ws_sales, ws_summary, ws_lookup]:
    for column_cells in ws.columns:
        column_letter = column_cells[0].column_letter
        max_length = max(
            len(str(cell.value or "")) for cell in column_cells
        )
        ws.column_dimensions[column_letter].width = min(max_length + 4, 30)

# Save workbook
wb.save(file_path)

print("Excel workbook created successfully!")
print("Saved to:", file_path)
```
